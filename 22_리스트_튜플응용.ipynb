```python
# 반복문으로 리스트의 요소를 모두 출력하기

a = [38,21,53,62,19]

for i in a:
    print(i)
```

```python
# 인덱스와 요소를 함께 출력하기

a = [38,21,53,62,19]

for index, value in enumerate(a):
    print(index,value)
    
i = (0,38)
```

```python
# while반복문으로 요소 출력하기

a = [38,21,53,62,19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 리스트의 가장 작은 수 구하기

a = [38,21,53,62,19]

smallest = a[0]

for i in a:
    if i < smallest:
        smallest = i
print(smallest)
```

```python
# 리스트의 가장 큰 수 구하기

a = [38,21,53,62,19]

largest = a[0]

for i in a:
    if i > largest:
        largest = i
print(largest)
```

```python
# 리스트 표현식 사용하기

# [식 for 변수 in 리스트]

a = [i for i in range(10)]
a
```

```python
a = []
for i in range(10):
    a.append(i)
print(a)
```

```python
c = []
for i in range(10):
    c.append(i+5)
print(c)
```

```python
[i + 5 for i in range(10)]
```

```python
b = []

for i in range(10):
    if i % 2 == 1:
        b.append(i+5)
print(b)
```

```python
[i+5 for i in range(10) if i % 2 == 1]
```

```python
c=[i*j for i in range(2,10) for j in range(1,10)]
print(c)
```

```python
c = []
for i in range(2,10):
    for j in range(1,10):
        c.append(i*j)
print(c)
```

```python
# 연습문제: 리스트에서 특정 요소만 뽑아내기

a = ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']
b = [i for i in a if len(i) == 5]
 
print(b)
```

```python
# 심사문제: 2의 거듭제곱 리스트 생성하기

x,y = map(int,input().split())

a = [2**i for i in range(x,y+1)]

a.pop(8)
a.pop(1)

print(a)
```
