```python
# 딕셔너리의 키에 접근하고 값 할당하기

lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux['health'])
print(lux['armor'])
```

```python
# 딕셔너리의 키에 값 할당하기

lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health'] = 2037
lux['mana'] = 1184

lux
```

```python
lux["mana_regen"] = 3.28

print(lux)
```
