```python
# 문제 1. 리스트 내장 함수

# 아래 리스트에 대해 내장 함수를 사용하여 다음을 구하시오.

scores = [85, 92, 78, 90, 88, 76, 95, 89]

# 1) 리스트의 길이 (len)
# 2) 최댓값 (max) 
# 3) 최솟값 (min)
# 4) 총합 (sum)
# 5) 평균 (sum과 len 활용)
# 6) 정렬된 새 리스트 반환 (sorted)
# 7) 내림차순 정렬 (sorted, reverse=True)

# A.
print(len(scores))
print(max(scores))
print(min(scores))
print(sum(scores))
print(sum(scores) / len(scores))
print(sorted(scores))
print(sorted(scores, reverse=True))



# 문제 2. 리스트와 문자열

# 다음 작업을 수행하시오.

# 1) 문자열 "Hello World"를 문자 리스트로 변환 (list)
# 2) 문자열 "a,b,c,d,e"를 쉼표로 분리하여 리스트로 변환 (split)
# 3) 리스트 ["Python", "is", "fun"]을 공백으로 연결하여 문자열로 변환 (join)

# A.
# 1)
x = "Hello World"
y = list(x)
print(y)

# 2)
x2 = "a,b,c,d,e"
y2 = x2.split(",")
print(y2)

# 3)
x3 = ["python","is","fun"]
y3 = " ".join(x3)
print(y3)


# 문제 3. 요소 추가 메서드

# 빈 리스트에서 시작하여 다음 작업을 순서대로 수행하시오.

# 1) append()로 10, 20, 30을 차례로 추가하시오.
# 2) insert()로 인덱스 1에 15를 삽입하시오.
# 3) insert()로 인덱스 0에 5를 삽입하시오.
# 4) extend()로 [40, 50]을 추가하시오.
# 5) 각 단계마다 리스트를 출력하시오.


# A.
x = []

[x.append(i) for i in range(10,31,10)]
print(x)

x.insert(1,15)
print(x)

x.insert(0,5)
print(x)

x.extend([40,50])
print(x)



# 문제 4. 요소 제거 메서드

# 아래 리스트에서 다음 작업을 수행하시오.

items = ["a", "b", "c", "b", "d", "b", "e"]

# 1) remove()로 첫 번째 "b"를 제거하시오.
# 2) pop()으로 마지막 요소를 제거하고, 제거된 값을 출력하시오.
# 3) pop(0)으로 첫 번째 요소를 제거하시오.
# 4) clear()로 모든 요소를 제거하시오.
# 5) 각 단계마다 리스트를 출력하시오.

# A.
items.remove("b")
print(items)

x = items.pop()
print(items)
print(x)

items.pop(0)
print(items)

items.clear()
print(items)




# 문제 5. 검색 및 카운트 메서드

# 아래 리스트에 대해 다음을 수행하시오.

data = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# 1) index()로 처음 나오는 5의 인덱스를 찾으시오.
# 2) count()로 5가 몇 번 나오는지 세시오.
# 3) count()로 1이 몇 번 나오는지 세시오.

# A.
print(data.index(5))
print(data.count(5))
print(data.count(1))



# 문제 6. 정렬 메서드

# 다음 작업을 수행하시오.

numbers = [64, 34, 25, 12, 22, 11, 90]
words = ["banana", "Apple", "cherry", "date"]

# 1) numbers를 sort()로 오름차순 정렬하시오.
# 2) numbers를 sort(reverse=True)로 내림차순 정렬하시오.
# 3) words를 알파벳순으로 정렬하시오.
# 4) reverse()로 리스트 순서를 뒤집으시오.


# A.
numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)

words.sort()
print(words)

words.reverse()
print(words)




# 문제 7. 튜플 생성과 기본 구조

# 다음 조건을 만족하는 프로그램을 작성하시오.

# 1) 정수 1, 2, 3, 4, 5를 담은 튜플을 생성하시오.
# 2) 문자열 "red", "green", "blue"를 담은 튜플을 생성하시오.
# 3) 요소가 하나뿐인 튜플을 올바르게 생성하시오. 
# 4) 빈 튜플을 두 가지 방법((), tuple())으로 생성하시오.
# 5) 괄호 없이 튜플을 생성하시오. 
# 6) 각 튜플의 길이(len)와 자료형(type)을 출력하시오.

# A.
x = (1,2,3,4,5)

c = ("red","green","blue")

o = (10,)

y1 = ()
y2 = tuple()

z = 10,20,30

print(len(x), type(x))
print(len(c), type(c))
print(len(o), type(o))
print(len(y1), type(y1))
print(len(y2), type(y2))
print(len(z), type(z))




# 문제 8. 튜플 생성과 기본 구조

# 다음 조건을 만족하는 프로그램을 작성하시오.

# 1) 정수 1, 2, 3, 4, 5를 담은 튜플을 생성하시오.
# 2) 문자열 "red", "green", "blue"를 담은 튜플을 생성하시오.
# 3) 요소가 하나뿐인 튜플을 올바르게 생성하시오. (주의: 쉼표 필요)
# 4) 빈 튜플을 두 가지 방법((), tuple())으로 생성하시오.
# 5) 괄호 없이 튜플을 생성하시오. (패킹)
# 6) 각 튜플의 길이(len)와 자료형(type)을 출력하시오.

# A.
t1 = (1, 2, 3, 4, 5)
t2 = ("red", "green", "blue")
t3 = (10,)
t4 = ()
t5 = tuple()
t6 = 1, 2, 3

for t in [t1, t2, t3, t4, t5, t6]:
    print(t)
    print(len(t))
    print(type(t))



# 문제 9. 튜플 인덱싱

# 아래 튜플이 주어졌을 때, 다음을 출력하시오.

colors = ("빨강", "주황", "노랑", "초록", "파랑", "남색", "보라")

# 1) 첫 번째 요소
# 2) 마지막 요소 (음수 인덱스 사용)
# 3) 세 번째 요소
# 4) 뒤에서 세 번째 요소
# 5) 인덱스 4에 해당하는 요소

# A.
print(colors[0])
print(colors[-1])
print(colors[2])
print(colors[-3])
print(colors[4])




# 문제 10. 딕셔너리 기본 개념
# 다음 질문에 답하시오.

# 1. 딕셔너리는 무엇으로 구성되어 있는가?
# 2. 딕셔너리와 리스트의 가장 큰 차이점은 무엇인가?
# 3. 다음 중 딕셔너리의 Key로 사용할 수 없는 것은? (이유도 함께 설명)
#    - a) 숫자 (1, 2, 3)
#    - b) 문자열 ("name", "age")
#    - c) 리스트 ([1, 2, 3])
#    - d) 튜플 ((1, 2, 3))

# A.
# 1. 딕셔너리는 key와 value 한 쌍으로 구성되어 있다.
# 2. 딕셔너리는 키로 데이터를 찾고 관리하며 리스트는 숫자로 순서대로 데이터를 찾고 관리한다.
# 3. -c) 리스트
#    딕셔너리의 키는 불변형 객체만 사용할 수 있다. 리스트는 가변형 객체이다.


# 문제 11. 코드 읽기
# 다음 코드의 실행하시오

student = {'name': '김철수', 'age': 20, 'major': '컴퓨터공학'}
print(student['name'])
print(student['age'])


# A.
김철수
20





# 문제 12. 딕셔너리 생성 및 조작
# 다음 조건에 맞는 딕셔너리를 생성하고 조작하시오.


# 1. 다음 정보를 담는 딕셔너리를 만드시오
#    - 이름: 본인 이름
#    - 나이: 본인 나이
#    - 취미: 본인 취미 (리스트로 여러 개)
#    - 거주지: 본인 거주지

# 2. 위 딕셔너리에 '전공' 항목을 추가하시오

# 3. '취미' 항목에 새로운 취미를 추가하시오

# 4. '나이' 항목의 값을 1 증가시키시오

# 5. 모든 key와 value를 출력하시오


# A.
me = {
    '- 이름': '시승완',
    '- 나이': 31,
    '- 취미': ['게임','그림'],
    '- 거주지': '대구'
}

me['전공'] = '없음'

me['취미'].append('웹서핑')

me['나이'] += 1

for key,value in me.item():
    print(key,':',value)
   


# 문제 13. 딕셔너리 함수 활용
# 다음 딕셔너리가 주어졌을 때, 각 요구사항을 수행하시오.


fruits = {
    'apple': 1000,
    'banana': 1500,
    'orange': 2000,
    'grape': 3000
}

# 1. 모든 과일 이름(key)을 리스트로 출력하시오

# 2. 모든 가격(value)을 리스트로 출력하시오

# 3. 'banana'의 가격을 출력하시오

# 4. 'melon'이 딕셔너리에 있는지 확인하시오

# 5. 'grape'를 pop() 함수를 사용하여 삭제하고, 삭제된 값을 출력하시오


# A.
print(list(fruits.keys()))
print(list(fruits.values()))
print(fruits['banana'])
print('melon' in fruits)

x = fruits.pop('grape')

print(x)
print(fruits)



# 문제 14. 성적 관리 프로그램
# 학생들의 성적을 관리하는 프로그램을 작성하시오.

# 다음 학생들의 성적 정보를 딕셔너리로 저장하시오
# 학생1: 이름=김민수, 국어=85, 영어=90, 수학=78
# 학생2: 이름=이영희, 국어=92, 영어=88, 수학=95
# 학생3: 이름=박철수, 국어=78, 영어=85, 수학=82

# 요구사항:
# 1. 각 학생의 총점과 평균을 계산하여 딕셔너리에 추가
# 2. 각 학생의 정보를 보기 좋게 출력



# **출력 예시:**

# === 학생 성적 관리 ===
# 이름: 김민수
# 국어: 85점, 영어: 90점, 수학: 78점
# 총점: 253점, 평균: 84.33점
# ---
# 이름: 이영희
# 국어: 92점, 영어: 88점, 수학: 95점
# 총점: 275점, 평균: 91.67점
# ---
# 이름: 박철수
# 국어: 78점, 영어: 85점, 수학: 82점
# 총점: 245점, 평균: 81.67점
# ---

# A.
s = [
    {'이름': '김민수','국어': 85,'영어': 90,'수학': 78},
    {'이름': '이영희','국어': 92,'영어': 88,'수학': 95},
    {'이름': '박철수','국어': 78,'영어': 85,'수학': 82}
]
for student in s:
    total = student['국어']+student['영어']+student['수학']
    average = total/3

    student['총점'] = total
    student['평균'] = average

print("=== 학생 성적 관리 ===")

for student in s:
    print(f"이름: {student['이름']}")
    print(f"국어: {student['국어']}점, 영어: {student['영어']}점, 수학: {student['수학']}점")
    print(f"총점: {student['총점']}점, 평균: {round(student['평균'], 2)}점")
    print("---")




# 문제 15. 단어 카운터
# 문자열에서 각 단어가 몇 번 등장하는지 세는 프로그램을 작성하시오.

text = "python is great python is fun python is powerful"

# 요구사항:
# 1. 문자열을 공백으로 분리하여 단어 리스트 만들기
# 2. 각 단어의 등장 횟수 세기
# 3. 결과를 출력



# **출력 예시:**

# python: 3번
# is: 3번
# great: 1번
# fun: 1번
# powerful: 1번

# A.
x = text.split()

counts = {}
for i in x:
    if i in counts:
        counts[i] += 1
    else:
        counts[i] = 1

for i,c in counts.items():
    print(f'{i}: {c}번')



# 문제 16. 중첩 딕셔너리
# 다음과 같은 중첩 딕셔너리를 다루는 문제를 해결하시오.


school = {
    '1반': {
        '김철수': {'국어': 85, '영어': 90, '수학': 78},
        '이영희': {'국어': 92, '영어': 88, '수학': 95}
    },
    '2반': {
        '박민수': {'국어': 88, '영어': 85, '수학': 90},
        '최지영': {'국어': 95, '영어': 92, '수학': 88}
    }
}

# 요구사항:
# 1. '1반' 김철수의 수학 점수를 출력하시오
# 2. '2반' 박민수의 평균 점수를 계산하여 출력하시오


# A.
print(school['1반']['김철수']['수학'])

x = school['2반']['박민수']

average = round(sum(x.values())/len(x))

print(f'{average}')




# 문제 17-1.
school = [
        {'이름': '김철수', '국어': 85, '영어': 90, '수학': 78},
        {'이름': '이영희', '국어': 92, '영어': 88, '수학': 95}, 
        {'이름': '박민수', '국어': 88, '영어': 85, '수학': 90},
        {'이름': '최지영', '국어': 95, '영어': 92, '수학': 88}
]
# 요구사항:
# 1. 최지영의 국어 점수를 출력하시오
# 2. 이영희의 평균 점수를 계산하여 출력하시오


# A.
print(school[3]["국어"])

student = school[1]

average = round((student['국어'] + student['영어'] + student['수학'])/3)

print(f'{average}')





# 문제 18. 다음 코드의 실행 결과를 예상하시오.

# 1)
s1 = set("Hello")
print(s1)

# 2)
s2 = set([1, 2, 3, 2, 1, 3, 4])
print(s2)

# 3)
s3 = {}
print(type(s3))

# A.
# {'H', 'e', 'l', 'o'}

# {'1,2,3,4}

# <class 'dict'>




# 17. 유닛23 지뢰찾기 문제를 풀어주세요


col,row = map(int, input().split())

# 문제에서는 표준 입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그 다음 줄부터 리스트의 요소로 들어갈 문자가 입력됩니다.
# 라고 했으니 co1(가로), row(세로) 순서로 넣어야 한다.


matrix = []
for i in range(row):
    matrix.append(list(input()))
    
# 2차원 리스트 가로 세로 입력값을 만드는 코드
# 3 3 입력시 col = 3, row = 3
# list(intput())은 지뢰 역활을 하는 요소를 담당하는 문자열을 리스트로 변환한다.
# ..*..을 ['.','.','*','.','.']로 변환


for y in range(row):
    for x in range(col):
        if matrix[y][x] == '*':
            continue
                
# 위에 작성한 문자열을 리스트로 변환했기에 matrix[y][x]는 y번째 행, x번쨰 열에 있는 문자를 뜻한다.
# if matrix[y][x] == '*': continue는 현재 칸이 지뢰'*'라면 코드를 실행하지 않고 다음 칸으로 이동한다는 뜻이다.

        
        count = 0
        
        for a in range(-1, 2):
            for b in range(-1, 2):
                c = y + a
                d = x + b

# count = 0 는 현재 칸의 주변 지뢰 개수를 뜻한다.
# 이는 현재 칸을 검사하는 반복문 for x in range(col): 안에 있어야한다.
# 그래야만 새로운 칸으로 이동할 때마다 지뢰 개수를 0부터 다시 센다.
# range(-1,2)는 -1에서 시작해서 2 직전까지 출력한다.(-1,0,1)
# 그래서 현재 칸을 기준으로 행렬을 각각 -1,0,1 만큼 이동하며 주변을 검사할 수 있으며 이는 총 9칸을 검사할 수 있다.


                if 0 <= c < row and 0 <= d < col:

# 우상단이나 좌상단같이 끝 칸에는 칸이 없는 경우가 있다.
# 그러므로 현재 좌표값이 정상인지 확인하는 코드를 넣어야한다.
# 0 <= c < row 행이 정상 범위인지. 0 <= d < col 열이 정상 범위인지.
# 음수를 막는 이유는 matrix[-1]이 리스트의 마지막 요소을 뜻하기에 반대쪽 지뢰까지 잘못 셀 수 있기 때문이다.

                    
                    if matrix[c][d] == '*':
                        count += 1     
                        
        matrix[y][x] = count

# 주변 위치에 지뢰가 있다면 count를 1 증가시킨다.

for i in matrix:
    for e in i:
        print(f'{e}', end='')
    print()
```
