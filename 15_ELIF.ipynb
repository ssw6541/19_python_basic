```python
# if 조걱식:
#     수행할코드
# elif 조건식:
#     수행할코드

x = 20

if x == 10:
    print('10입니다.')
elif x == 20:
    print('20입니다.')
```

```python
# if, elif, else를 모두 사용하기

x = 30

if x == 10:
    print('10입니다.')
elif x == 20:
    print('20입니다.')
else:
    print('10도 20도 아닙니다.')
    
```

```python
x = 30

if x == 10:
    print('10입니다.')
else:
    print('10도 20도 아닙니다.')
elif x == 20:
    print('20입니다.')
    
```

```python
# 음료수 자판기 만들기

button = int(input())

if button == 1:
    print('콜라')
elif button == 2:
    print('사이다')
elif button == 3:
    print('환타')
else:
    print('제공하지 않는 메뉴입니다.')
    
```

```python
x = int(input())

if 11<=x<=20:
    print('11~20')
elif 21<=x<=30:
    print('21~30')
else:
    print('아무것도 해당하지 않음')
```

```python
age = int(input())
balance = 9000 # 교통카드 잔액

if 7<=age<=12:
   balance -= 650
elif 13<=age<=18:
    balance -= 1050
elif 19<=age:
    balance -= 1250

print(balance)
```
