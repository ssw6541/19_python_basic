```python
# 문자열의 왼쪽, 오른쪽, 가운데 정렬하기

print('python'.ljust(10)) # 글자6 + 공백4
print('python'.rjust(10)) # 공백4 + 글자6
print('python'.center(10)) # 공백2 + 글자6 + 공백2
```

```python
# 심사문제: 높은 가격순으로 출력하기

x = input().split(';')

x = list(map(int,x))
x.sort(reverse=True)

for i in x:
    print('{0:9,}'.format(i))
```

```python
x = input().split(';')

x = list(map(int,x))
x.sort(reverse=True)

for i in x:
    print(f'{i:9,}')
```
