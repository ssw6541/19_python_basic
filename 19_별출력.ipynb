```python
# 중첩 루프 사용하기

for i in range(5):
    
    for j in range(5):
        print('j:',j,sep='',end=' ')
        
    print('i:',i,'\\n',sep='')
```

```python
# 사각형으로 별 출력하기

for i in range(5):
    for j in range(5):
        print('*',end='   ')
    print()
```

```python
# 사각형 모양 바꾸기

for i in range(3):
    for j in range(7):
        print('*',end='')
    print()
```

```python
# 계단식으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j <= i:
            print('*',end='')
    print()
```

```python
# 대각선으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j == i:
            print('*',end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print('*',end='')
        else:
            print('    ',end='')
    print()
```

```python
# 연습문제: 역삼각형 모양으로 별 출력하기

for i in range(5):
    for j in range(5):
        if j < i:
            print(' ', end='')
        else:
            print('*', end='')
    print()
```

```python
# 심사문제: 산 모양으로 별 출력하기

x = int(input())

for i in range(x):
    for j in range(x-i-1):
        print(' ',end='')
    for j in range(2*i+1):
        print('*',end='')
    print()

```

```python
x = int(input())

for i in range(x):
    print(' '*(x-i-1)+'*'*(2*i+1))
```
