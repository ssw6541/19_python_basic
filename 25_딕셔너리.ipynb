```python
# 딕셔너리에 키와 값 정하기
# {키:기본값}
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40} 
x.setdefault('e')
x
```

```python
# 딕셔너리에서 키의 값 수정하기

x = {'a':10,'b':20,'c':30,'d':40}
x.update(a=90)
x
```

```python
x.update(e=50)
x
```

```python
x.update(a=900,f=60)
x
```

```python
# 딕셔너리에서 키-값 쌍 삭제하기

x = {'a':10,'b':20,'c':30,'d':40}
x.pop('a')
x
```

```python
x = {'a':10,'b':20,'c':30,'d':40} # python 3.6 이상에서는 마지막 키-값을 삭제한다.
x.popitem()
x
```

```python
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.clear()
x
```

```python
# 딕셔너리에서 키 값을 가져오기

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.get('a')
```

```python
x.get('z')
```

```python
x.get('z',0)
```

```python
# 딕셔너리에서 키-값 쌍을 모두 가져오기

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.items()
```

```python
x.keys()
```

```python
x.values()
```

```python
# 반복문으로 키-값 모두 출력하기

x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

for i in x:
    print(i, end=' ')
```

```python
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

for i in x.items():
    print(i)
```

```python
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

for x, y in x.items():
    print(x,y)
```
