```python
# 값을 여러개 출력하기

print(1, 2, 3)
```

```python
print('hello','python')
```

```python
print(1, 2, 3, sep=', ')
```

```python
print(4, 5, 6, sep=',')
```

```python
print('hello', 'python', sep='')
```

```python
print('1920', '1080', sep='x')
```

```python
# 줄바꿈 활용하기

print(1, 2, 3)
```

```python
print(1, 2, 3, sep='\n')    # 개행 문자 (\n)
```

```python
print('1\n2\n3')    #\n: 다음 줄로 이동하며 개행이라고도 부릅니다.

                    #\t: 탭 문자, 키보드의 Tab 키와 같으며 여러 칸을 띄웁니다.

                    #\\: \ 문자 자체를 출력할 때는 \를 두 번 써야 합니다.
```

```python
# end 한줄에 여러 개의 값을 출력

print(1,end='')
print(2,end='')
print(3)
```

```python
print(1,end=' ')    #공백을 주면 공백이 생긴다
print(2,end=' ')
print(3)
```
